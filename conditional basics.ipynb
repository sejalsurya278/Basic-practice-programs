```python
#nested if
score=89
attendance=8

if score>=70:
    print("student passed the exam")

    if attendance>=90:
        print("student also has an excellent performance in attendance")
    else:
        print("student can improve performance in attendance")
else:
    print("student failed the exam")
```

```python
#nested loop

adj=["red","big","tasty"]
fruits=["apple","banana","cherry"]
  
for x in adj:
    for y in fruits:
       print(x,y)
```
