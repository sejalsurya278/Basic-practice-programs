```python
import math

try:
    # Input coefficients
    a = float(input("Enter a: "))
    b = float(input("Enter b: "))
    c = float(input("Enter c: "))

    # Check if quadratic
    if a == 0:
        print("Not a quadratic equation")
    else:
        d = b**2 - 4*a*c   # discriminant

        # Square root (may cause ValueError if d < 0)
        root = math.sqrt(d)

        # Roots
        r1 = (-b + root) / (2*a)
        r2 = (-b - root) / (2*a)

        print("Roots are:", r1, "and", r2)

except ValueError:
    print("ValueError: Invalid input or negative discriminant")

except TypeError:
    print("TypeError: Wrong data type used")

except NameError:
    print("NameError: Variable not defined")
```

```python
#recursive 

def print_numbers(n):

    if n==0 :

        return

    print_numbers(n-1)
    print(n)

n=int(input("enter n"))
print_numbers(n)
```
