```python
#12 program to get highest marks of student
studentdict={
    "sejal":100,
    "arnav":67,
    "om":56,
    "shiv":99
  }
#highest student mein we will save the max of 
highest_student=max(studentdict,key=studentdict.get)
print("highest marks student is:",highest_student)
print("highest marks",studentdict[highest_student])
```

```python
#13 python program to count the frequency of each element in alist using a dictionary

numbers=[10,20,10,40,20,20,50,30]
frequency={}

for num in numbers:
    if num in frequency:
        frequency[num]=frequency[num]+1
    else:
        frequency[num]=1
print("frequency of elements:",frequency)
```

```python
#14 to sort dictionary by its values

studentdict={
    "sejal":100,
    "arnav":67,
    "om":56,
    "shiv":99
  }

sorted_dict=dict(sorted(studentdict.items(),key=lambda item:item[1]))
print("Dictionary sorted by value:",sorted_dict)
```

```python
#15 for descending order

studentdict={
    "sejal":100,
    "arnav":67,
    "om":56,
    "shiv":99
  }

sorted_dict=dict(sorted(studentdict.items(),key=lambda item:item[1],
                       reverse=True))
print("Dictionary sorted by value:",sorted_dict)
```
