```python
#creating a series
import pandas as pd
a=[3,4,5]

b=pd.Series(a)
print(b)
```

```python
#renaming their indexes
import pandas as pd
a=[3,7,8,9]

b=pd.Series(a,index=["a","b","c","d"])
print(b)
```

```python
#creating a dictionary 
import pandas as pd
a={
    "day1":59,
    "day2":78,
    "day3":88
}

b=pd.Series(a)
print(b)
```

```python
#cleansing the data dropna
import pandas as pd

a={
    "a1":[1,2,3,None],
    "a2":[None,5,6,7],
    "a3":[9,6,None,2,]
}

c=pd.DataFrame(a)

print("\nOriginal Data",c)
c.dropna(inplace=True)
print("\nAfter cleansing the data",c)
```

```python
#filling 0 in the places where there is none fillna
import pandas as pd

a={
    "a1":[1,2,3,None],
    "a2":[None,5,6,7],
    "a3":[9,6,None,2,]
}

c=pd.DataFrame(a)

print("\nOriginal Data",c)

#c.fillna(0,inplace=True)

c_filling=c.fillna(0)
print("After using Fillna",c_filling)
```

```python
import pandas as pd

a={
     "a1":[9,2,3,None,8],
    "a2":[None,5,6,7,8],
    "a3":[9,6,None,2,8]
}

c=pd.DataFrame(a)
print("\nOriginal Data",c)

c.fillna(c.mean(),inplace=True)
print("\nafter filling with the mean",c)
```

```python
#renaming
import pandas as pd

a={
     "a1":[9,2,3,None,8],
    "a2":[None,5,6,7,8],
    "a3":[9,6,None,2,8]
}

c=pd.DataFrame(a)

print("\nOriginal data",c)

c.rename(columns={"a1":"age", "a2":"place","a3":"gender"},inplace=True)
print(c)
```
