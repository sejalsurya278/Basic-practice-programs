```python
# tuple questions practice


#write program to print 4th element from first and 4th element from last in a tuple

numbers=(10,20,30,40,50,60,70,80,90)
print("\nTuple:",numbers)

print("\n4th element from start:", numbers[3])
print("\n4th element from last:", numbers[-4])
```

```python
#program to check whether an element exists in tupple or not

numbers=(10,20,30,40,50,60)
element=int(input("\nenter a number to check:"))

if element in numbers:
    print("\nelement exists in tuple")
else: 
    print("does not exist")
```

```python
#convert list into tuple
number_list=[10,20,30,40,50]
number_tuple=tuple(number_list)

print("\nnumber list is:",number_list)
print("\nnumber tuple after converting is:",number_tuple)
```

```python
#program to find index of an element in tuple

numbers=(10,20,40,60,33,29)
element=int(input("\nenter to a number to find its index"))

if element in numbers:
    print("index of number is",numbers.index(element))
else:
    print("\nthis number does not exist in this tuple")
```

```python
#program to replace last values of tuple in list to 100

number=[(10,50,88),(67,89,67),(34,89,23)]
new_list=[]

for tup in number:
    new_tup=tup[:-1]+(100,)
    new_list.append(new_tup)
print("original list:",number)
print("updated list:",new_list)
```

```python
#program to find the maximum and minimum of tuple

numbers=(10,20,40,60,33,29)
maximum=max(numbers)
minimum=min(numbers)

print("\nmax number is:",maximum)
print("\nmin number is:",minimum)
```

```python
#program to swap 1st and last element in tuple
numbers=(10,20,40,60,33,29)
new_tuple= (numbers[-1],) + numbers[1:-1] + (numbers[0],)

print("\noriginal tuple:",numbers)
print("\nafter swapping the numbers:",new_tuple)
```
