```python
#dictionary programs

#to add a new key and value in dictionary
dict={101:1,102:2,103:4}

dict[176]=30
print("\nafter adding new key and value",dict)
```

```python
#program to merge 3 dictionaries

dict1={101:1,102:2,103:4}
dict2={109:1,102:1,103:3}
dict3={107:1,182:2,173:4}

new_dict={}
new_dict.update(dict1)
new_dict.update(dict2)
new_dict.update(dict3)
print("\nMerged dictionary is:",new_dict)
```

```python
#program to check if key already exists or not
dict={101:1,102:2,103:4}

element=int(input("\nenter a key to check if it exists or not"))

if element in dict:
    print("\nthe key already exists")
else: 
    print("\nkey does not exist")
```

```python
#program to print key and value separately

#imp syntax use of .values to get value but can use keys directly also to get key an value both we u items
dict={101:1,102:2,103:4}

print("Keys:")
for key in dict:
    print(key)
    
print("value")
for value in dict.values():
    print(value)

print("keys and value :")
for key,value in dict.items():
    print(key,value)
```

```python
#prepare dictionary where in keys are numbers between 1-15 and values are square of the keys

dict={}

for i in range(1,16):
     dict[i]=i**2

print("\ndictionary:",dict)
```

```python
#program to sum all the value in dictionary

dict={"a":101,"b":102,"c":103,"d":104}
total=0

for value in dict.values():
    total=total+value
print("\nsum of values:",total)
```
