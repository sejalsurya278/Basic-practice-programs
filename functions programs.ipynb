```python
#1:python program to accept 3 numbers and display their sum using function

x=int(input("enter x"))
y=int(input("enter y"))
z=int(input("enter z"))

def cal_sum(x,y,z):
    sum=x+y+z
    return sum

s=cal_sum(x,y,z)
print(s)
```

```python
#2:program to receive 3 inputs to get their sum and product calculated by a user defined function

def cal_sumprod(x,y,z):
    pp=x*y*z
    ss=x+y+z

    return pp,ss

x=int(input("enter x"))
y=int(input("enter y"))
z=int(input("enter z"))

s,p = cal_sumprod(x,y,z)
print(s,p)
```

```python
#3: accept numbers from user and display prime factors of that number



def prime_factors(n):
    for i in range(2, n + 1):
        while n % i == 0:
            print(i, end=" ")
            n = n // i

n = int(input("Enter a number: "))
prime_factors(n)
```
